### Recipe Suggester Agent

```python
# Import the necessary libraries
from dotenv import load_dotenv
load_dotenv(override=True)

from agents import Agent, Runner, trace
```

```python
# Make an agent with name, instructions, model
agent = Agent(name="Recipe Suggester", instructions="You are a recipe suggester. You suggest recipes based on the ingredients people have.", model="gpt-4o-mini")
```

```python
# Run the suggester with Runner.run(agent, prompt) then point final_output, then wrap in in a trace
with trace("Suggesting a recipe"):
    result = await Runner.run(agent, "I have chicken, rice, and broccoli. What can I make for dinner?")
    print(result.final_output)
```
