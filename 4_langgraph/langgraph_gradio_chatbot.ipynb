```python
# Import necessary libraries
from typing import Annotated
from langgraph.graph import StateGraph, START, END
from langgraph.graph.message import add_messages
from dotenv import load_dotenv
from IPython.display import display, Image
import gradio as gr
from pydantic import BaseModel
from langchain_openai import ChatOpenAI
import random
```

```python
# Some useful constants
nouns = ["Cabbages", "Unicorns", "Pineapples", "Kangaroos", "Marshmallows"]
adjectives = ["Fluffy", "Sparkling", "Giant", "Tiny", "Invisible"]
```

```python
# Load environment variables from .env file
load_dotenv(override=True)
```

```python
# Define a function to shout text
def shout(text: Annotated[str, "Something to be shouted"]) -> str:
    print(text.upper())
    return text.upper()

shout("Hello, world!")
```

### Step 1: Define the State object

```python
# Define a Pydantic model for the state
class State(BaseModel):
    messages: Annotated[list, add_messages]
```

### Step 2: Start the Graph Builder with this State class

```python
# Create a state graph using the State model
graph_builder = StateGraph(State)
```

### Step 3: Create a Node

```python
# Define a function for the first node in the state graph
def our_first_node(old_state: State) -> State:

    reply = f"{random.choice(nouns)} are {random.choice(adjectives)}"
    messages = [{"role": "assistant", "content": reply}]

    new_state = State(messages=messages)

    return new_state

graph_builder.add_node("first_node", our_first_node)
```

### Step 4: Create Edges

```python
# Add edges to the state graph
graph_builder.add_edge(START, "first_node")
graph_builder.add_edge("first_node", END)
```

### Step 5: Compile the Graph

```python
# Compile the state graph into a runnable graph
graph = graph_builder.compile()
```

```python
# Display the graph
display(Image(graph.get_graph().draw_mermaid_png()))
```

```python
# Define a function to handle chat interactions
def chat(user_input: str, history):
    message = {"role": "user", "content": user_input}
    messages = history + [message]
    state = State(messages=messages)
    result = graph.invoke(state)
    print(result)
    return result["messages"][-1].content

gr.ChatInterface(chat).launch(share=True)
```

### Let's add LLM

```python
# Define a Pydantic model for the state
class State(BaseModel):
    messages: Annotated[list, add_messages]
```

```python
# Create a state graph using the State model
graph_builder = StateGraph(State)
```

```python
# Define a function for the chatbot node in the state graph
llm = ChatOpenAI(model="gpt-4o-mini")

def chatbot_node(old_state: State) -> State:
    response = llm.invoke(old_state.messages)
    new_state = State(messages=[response])
    return new_state

graph_builder.add_node("chatbot", chatbot_node)
```

```python
# Add edges to the state graph
graph_builder.add_edge(START, "chatbot")
graph_builder.add_edge("chatbot", END)
```

```python
# Compile the state graph into a runnable graph
graph = graph_builder.compile()
display(Image(graph.get_graph().draw_mermaid_png()))
```

```python
# Define a function to handle chat interactions
def chat(user_input: str, history): 
    initial_state = State(messages= history + [{"role": "user", "content": user_input}])
    result = graph.invoke(initial_state)
    print(result)
    return result["messages"][-1].content

gr.ChatInterface(chat).launch(share=True)
```
